```python
import numpy as np
import matplotlib.pyplot as plt 

P = [0.2, 0.5, 0.8, "(1/n)^(1/2)"]
```

```python

fileNames, algos = ["MaxBranche", "MaxBrancheBound", "MaxBrancheBoundV2", "MaxBrancheBoundV3", "MaxBrancheBoundV4",
                    "MaxCouplage", "MaxGloutton"], \
                    ["de branchement simple", "de branch and bound", "de branch and bound amélioré", 
                     "de branch and bound avec degMax", "de branch and bound avec degMax amélioré", "de couplage", "glouton"]


def f(fileNameStart, algo) :

    X, Y = [], []
    f = open(fileNameStart + "2.txt", "r")
    tempX, tempY = [], []
    for l in f : 
        u, v = l.split(",")
        tempX.append(float(u))
        tempY.append(int(v))
    X.append(tempX)
    Y.append(tempY)

    f = open(fileNameStart + "5.txt", "r")
    tempX, tempY = [], []
    for l in f : 
        u, v = l.split(",")
        tempX.append(float(u))
        tempY.append(int(v))
    X.append(tempX)
    Y.append(tempY)

    f = open(fileNameStart + "8.txt", "r")
    tempX, tempY = [], []
    for l in f : 
        u, v = l.split(",")
        tempX.append(float(u))
        tempY.append(int(v))
    X.append(tempX)
    Y.append(tempY)


    plt.suptitle('Taille max du graphe en fonction du temps\n pour l\'algoritme ' + algo)
    
    for i in range(3) : 
        plt.plot(X[i], Y[i], label='p = ' + str(P[i]))
    
    plt.ylabel("Temps en s")
    plt.xlabel("Nombre maximal de sommets")

    legend = plt.legend(loc='best', shadow=True)

    # Put a nicer background color on the legend.


    plt.tight_layout()
    plt.show()

def fInv(fileNameStart, algo) :

    X, Y = [], []
    f = open(fileNameStart + "2.txt", "r")
    tempX, tempY = [], []
    for l in f : 
        u, v = l.split(",")
        tempX.append(float(u))
        tempY.append(int(v))
    X.append(tempX)
    Y.append(tempY)

    f = open(fileNameStart + "5.txt", "r")
    tempX, tempY = [], []
    for l in f : 
        u, v = l.split(",")
        tempX.append(float(u))
        tempY.append(int(v))
    X.append(tempX)
    Y.append(tempY)

    f = open(fileNameStart + "8.txt", "r")
    tempX, tempY = [], []
    for l in f : 
        u, v = l.split(",")
        tempX.append(float(u))
        tempY.append(int(v))
    X.append(tempX)
    Y.append(tempY)

    f = open(fileNameStart + "N.txt", "r")
    tempX, tempY = [], []
    for l in f : 
        u, v = l.split(",")
        tempX.append(float(u))
        tempY.append(int(v))
    X.append(tempX)
    Y.append(tempY)

    plt.suptitle('Taille max du graphe en fonction du temps\n pour l\'algoritme ' + algo)

    for i in range(4) : 
        plt.plot(Y[i], X[i], label='p = ' + str(P[i]))
    plt.ylabel("Temps en s")
    plt.xlabel("Nombre maximal de sommets")

    legend = plt.legend(loc='best', shadow=True)

    # Put a nicer background color on the legend.

    plt.tight_layout()
    plt.show()
```

```python

f(fileNames[-1], algos[-1])
f(fileNames[-2], algos[-2])
fInv(fileNames[0], algos[0])
fInv(fileNames[1], algos[1])
fInv(fileNames[2], algos[2])
fInv(fileNames[3], algos[3])
fInv(fileNames[4], algos[4])

```

```python

fileNames, algos = ["PerfBranch", "PerfBrancheBound", "PerfBrancheBoundV2", "PerfBrancheBoundV3", "PerfBrancheBoundV4",
                    "PerfCoup", "PerfGlout"], \
                    ["de branchement simple", "de branch and bound", "de branch and bound amélioré", 
                     "de branch and bound avec degMax", "de branch and bound avec degMax amélioré", "de couplage", "glouton"]

def fInv(fileNameStart, algo, k) :

    X, Y = [], []
    f = open(fileNameStart + "2.txt", "r")
    tempX, tempY = [], []
    for l in f : 
        u, v = l.split(",")
        tempX.append(int(u))
        tempY.append(float(v))
    X.append(tempX)
    Y.append(tempY)

    f = open(fileNameStart + "5.txt", "r")
    tempX, tempY = [], []
    for l in f : 
        u, v = l.split(",")
        tempX.append(int(u))
        tempY.append(float(v))
    X.append(tempX)
    Y.append(tempY)

    f = open(fileNameStart + "8.txt", "r")
    tempX, tempY = [], []
    for l in f : 
        
        u, v = l.split(",")
        tempX.append(int(u))
        tempY.append(float(v))
    X.append(tempX)
    Y.append(tempY)

    if (k == 4) : 
        f = open(fileNameStart + "N.txt", "r")
        tempX, tempY = [], []
        for l in f : 
            
            u, v = l.split(",")
            tempX.append(int(u))
            tempY.append(float(v))
        X.append(tempX)
        Y.append(tempY)

    plt.suptitle('Temps moyen de l\'algorithme ' + algo)

    for i in range(k) : 
        plt.plot(X[i], Y[i], label='p = ' + str(P[i]))
    plt.ylabel("Temps en s")
    plt.xlabel("Nombre maximal de sommets")

    legend = plt.legend(loc='best', shadow=True)

    # Put a nicer background color on the legend.

    plt.tight_layout()
    plt.show()
```

```python
fInv(fileNames[-1], algos[-1], 3)
fInv(fileNames[-2], algos[-2], 3)
fInv(fileNames[0], algos[0], 4)
fInv(fileNames[1], algos[1], 4)
fInv(fileNames[2], algos[2], 4)
fInv(fileNames[3], algos[3], 4)
fInv(fileNames[4], algos[4], 4)
```
