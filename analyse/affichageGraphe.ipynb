```python
import numpy as np
import matplotlib.pyplot as plt 

f = open("MaxGloutton2.txt", "r")

X, Y = [], []
for l in f : 
    u, v = l.split(",")
    X.append(float(u))
    Y.append(int(v))

plt.suptitle('Taille max du graphe en fonction du temps\n pour l\'algoritme gloutton pour p = 0,2.')

plt.subplot(1, 2, 1)
plt.plot(X, Y)
plt.xlabel("Temps en s")
plt.ylabel("Nombre maximal de sommets")

plt.subplot(1, 2, 2)
plt.loglog(X, Y)
plt.title("Échelle logarithmique")
plt.xlabel("Temps en s")

plt.tight_layout()
plt.show()
```
