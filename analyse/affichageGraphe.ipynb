```python
import numpy as np
import matplotlib.pyplot as plt 

f = open("ResultsCouplage8.txt", "r")

X, Y = [], []
for l in f : 
    u, v = l.split(",")
    X.append(u)
    Y.append(v)

"""
Sachant une probabilité de 0.8
"""
plt.plot(X, Y)
plt.show()
plt.loglog(X, Y)
plt.show()
```
